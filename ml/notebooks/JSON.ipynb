```python
import pandas as pd
```

```python
df = pd.read_csv('rawg_games_20k_with_images.csv')
```

```python
df.shape
```

```python
df_copy = df.copy()
```

```python
df_copy.columns.tolist()
```

```python
req_cols = ['id', 'name', 'released', 'rating', 'platforms', 'genres', 'tags','background_image']
df_req_cols = df_copy[req_cols]
```

```python
df_req_cols
```

```python
df_req_cols_copy = df_req_cols.copy()
```

```python
df_req_cols_copy
```

```python
df_req_cols['tags'].iloc[0]
```

```python
def extract_tags(tags_info):

      if pd.isna(tags_info):
        return ""

      separated_tags = str(tags_info).split('|')
      new_taglist = []

      for tag in separated_tags:
        current_tags = tag.strip().lower()

        if 'single' in current_tags or 'multi' in current_tags:
            new_taglist.append(tag.strip())
      return ', '.join(new_taglist)
df_req_cols_copy['tags'] = df_req_cols_copy['tags'].apply(extract_tags)
```
